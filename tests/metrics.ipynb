```python
import anndata
import numpy as np
import torch
from torchmetrics import R2Score, PearsonCorrCoef

# Creating a toy AnnData object
num_cells = 100
num_genes = 50

# Randomly generated gene expression data
X = np.random.rand(num_cells, num_genes)

# Creating the AnnData object
adata = anndata.AnnData(X)

# Simulating predicted and true expression values
y_true = torch.tensor(X, dtype=torch.float32)  # True gene expression
out_transformer = torch.tensor(np.random.rand(num_cells, num_genes), dtype=torch.float32)  # Model predictions

# Using torchmetrics to calculate the R² score
r2_genes = R2Score(num_outputs=num_genes, multioutput='raw_values')
r2_score_genes = r2_genes(out_transformer, y_true)

r2_cells = R2Score(num_outputs=num_cells, multioutput='raw_values')
r2_score_cells = r2_cells(torch.transpose(out_transformer, 0, 1), torch.transpose(y_true, 0, 1))

print(f"R² score (genes): {r2_score_genes, r2_score_genes.mean()}")
print(f"R² score (cells): {r2_score_cells.mean()}")
```

```python
pearson_corr = PearsonCorrCoef(num_outputs=num_genes)
pc = torch.mean(pearson_corr(out_transformer, y_true))
print(f'Pearson correlation: {pc}')
```
