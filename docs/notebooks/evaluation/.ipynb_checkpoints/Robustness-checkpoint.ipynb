```python
# Loop through each model and plot the data
models = {}
models['GNN'] = GNN
models['InterScale'] = InterScale

colors = ["blue", "green", "orange", "red", "purple"]

nr_classes = 8

fig, axs = plt.subplots(1, nr_classes, figsize=(12, 6), sharey=True)

for idx, (model_name, data) in enumerate(models.items()):
    R_values = list(data.keys())
    R_numeric = [int(R) for R in R_values]  # Convert R keys to numeric
    acc_values = [data[R]["test_acc"] for R in R_values]
    
    # Overall accuracy plot
    axs[0].plot(
        R_numeric,
        acc_values,
        marker="o",
        label=f"{model_name} (Overall)",
        color=colors[idx % len(colors)],
    )
    
    for class_id in range(nr_classes):
        class_str = f"test_f1/class_{class_id}"
        class1_values = [data[R][class_str] for R in R_values]
    
        # Class-specific accuracy plot
        axs[class_id].plot(
            R_numeric,
            class1_values,
            marker="s",
            label=f"{model_name} (Class {class_id})",
            color=colors[idx % len(colors)],
        )

# Add titles, labels, and legends
axs[0].set_title("Overall Accuracy")
axs[0].set_xlabel("R Value")
axs[0].set_ylabel("Accuracy")
axs[0].grid(True)
axs[0].legend()

for class_id in range(nr_classes):
    axs[class_id].set_title("Class-Specific Accuracy")
    axs[class_id].set_xlabel("R Value")
    axs[class_id].grid(True)
    axs[class_id].legend()

plt.tight_layout()
plt.show()
```
