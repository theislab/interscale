```python
# Extract data for the plot
model_names = list(models_results.keys())
metrics = ["train_f1/class_0", "train_f1/class_1", "train_f1/class_2", "train_f1/class_3", "train_f1/class_4", "train_f1/class_5", "train_f1/class_6", "train_f1/class_7", "train_f1/class_8"]
class_names = ['Acinar', 'Alpha', 'Beta', 'Ductal', 'Endocrine', 'Endothelial', 'Fibroblasts', 'Immune', 'Mast']
values = {metric: [models_results[model][metric] for model in model_names] for metric in metrics}

# Set up the bar plot
x = np.arange(len(model_names))
bar_width = 0.1
fig, ax = plt.subplots(figsize=(10, 6))

# Plot bars for each metric
for i, metric in enumerate(metrics):
    ax.bar(x + i * bar_width, values[metric], width=bar_width, label=metric)

# Add labels, title, and legend
ax.set_xlabel("Models", fontsize=14)
ax.set_ylabel("Scores", fontsize=14)
ax.set_title("F1 Node label scores", fontsize=16)
ax.set_xticks(x + bar_width)
ax.set_xticklabels(model_names, fontsize=12)
ax.legend(fontsize=12, labels = class_names, bbox_to_anchor=(1.5, 0.5), ncols = 2)

# Adjust layout and save the figure
plt.tight_layout()
plt.savefig("model_comparison_plot.png", dpi=300)
plt.show()
```
