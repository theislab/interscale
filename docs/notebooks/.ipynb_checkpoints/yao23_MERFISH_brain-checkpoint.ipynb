# Preprocess data

```python
adata.obs['x'] = adata.obs['center_x']
adata.obs['y'] = adata.obs['center_y']
```

```python
adata = sliding_windows(adata, 800, library_key = 'library_key',  overlap=0)
```

```python
cells_per_window = adata.obs.groupby('sliding_window').size()
print(f'Cells per window stats: Max: {cells_per_window.max()}, MIN: {cells_per_window.min()}, mean: {cells_per_window.mean()}, nr. of windows: {len(cells_per_window)}')
```

```python
import matplotlib.pyplot as plt

def plot_sliding_window_size_distribution(adata):
    # Get the size of each group
    group_sizes = adata.obs.groupby('sliding_window').size()
    
    # Plot the distribution
    plt.figure(figsize=(10, 6))
    plt.hist(group_sizes, bins=30, edgecolor='k', alpha=0.7)
    plt.title('Distribution of Group Sizes (sliding_window)', fontsize=16)
    plt.xlabel('Group Size', fontsize=14)
    plt.ylabel('Frequency', fontsize=14)
    plt.grid(True)
    plt.show()

plot_sliding_window_size_distribution(adata)
```
