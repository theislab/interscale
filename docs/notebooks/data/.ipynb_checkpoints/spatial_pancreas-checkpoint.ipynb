## Prepare data

Split data into windows

```python
adata.obs['x'] = adata.obs['CenterX_global_px']
adata.obs['y'] = adata.obs['CenterY_global_px']
```

```python
#sliding_window(adata, library_key = 'fov', window_size=2000, overlap=1000)
sliding_window(adata, library_key = 'fov', window_size=2000, overlap=0)
adata
```

```python
cells_per_window = adata.obs.groupby('sliding_window').size()
print(f'Cells per window stats: Max: {cells_per_window.max()}, MIN: {cells_per_window.min()}, mean: {cells_per_window.mean()}, nr. of windows: {len(cells_per_window)}')
```

```python
for sliding_window in ['sliding_window_1', 'sliding_window_2', 'sliding_window_3', 'sliding_window_4']:
    adata.obs[sliding_window] = adata.obs[sliding_window].astype("category")
    adata.obs[sliding_window]
```
